3. **Гіпотеза №3 - перевірка обсягу покупки та ціни (Quantity та Unit price)**:

В Mandalay вищий gross income, тому що покупці беруть більшу кількість товарів за один візит (Quantity) або купують дорожчі позиції (Unit price).

```python
# Кількість товарів у чеку
yangon_qty = df.query("City == 'Yangon'")['Quantity'].mean()
mandalay_qty = df.query("City == 'Mandalay'")['Quantity'].mean()

print(f"Середня кількість товарів: Yangon = {yangon_qty:.2f}, Mandalay = {mandalay_qty:.2f}")

# Ціна за одиницю товару
yangon_price = df.query("City == 'Yangon'")['Unit price'].mean()
mandalay_price = df.query("City == 'Mandalay'")['Unit price'].mean()

print(f"Середня ціна товару: Yangon = {yangon_price:.2f}, Mandalay = {mandalay_price:.2f}")
```
