Гіпотеза 2.
    Чи існує взаємозв'язок між обраним способом оплати (Cash, Credit card, Ewallet), середньою кількістю товарів у чеку (Quantity) та середнім значенням рейтингу задоволеності клієнта (Rating)?

```python
deversification_by_payment = df.groupby('Payment').agg(
    avg_quantity=('Quantity', 'mean'),
    avg_rating=('Rating', 'mean'),
    total_transactions=('Quantity', 'count')
).round(2)
```

```python
deversification_by_payment
```

Спосіб оплати практично не впливає ані на середнє наповнення чека товарів (~5.5 шт.), ані на рівень задоволеності клієнтів (~7.0 балів).

**Рекомендації для бізнесу**
    Попри те, що **Credit card** має найвищий рейтинг **(7.00)**, цей метод є найменш популярним за кількістю транзакцій **(311 проти ~345 у Cash та Ewallet)**.
